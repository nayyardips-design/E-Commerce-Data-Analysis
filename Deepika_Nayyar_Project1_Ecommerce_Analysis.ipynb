## Q3. Missing values

```python
def missing_values(df):
    out = pd.DataFrame({'missing_count': df.isna().sum(),'missing_pct':df.isna().mean()*100})
    return out[out.missing_count>0].sort_values(by = 'missing_count',ascending = False)

for name,df in {'orders':orders,'items':items,'products':products,'customers':customers}.items():
    print(f'\n {name.upper()}')
    display(missing_values(df))
```

**Cleaning decisions:** `order_approved_at`, carrier date and delivered-customer date are legitimately missing for some orders at different stages, so they should not automatically be replaced. Product category has 610 missing values and is filled as `unknown`; small numeric product-data gaps are filled with the column median for analysis.

# Part 2. Data Cleaning

## Q5. Clean date columns and find earliest/latest order dates

```python
order_date_cols=['order_purchase_timestamp','order_approved_at','order_delivered_carrier_date','order_delivered_customer_date','order_estimated_delivery_date']
for c in order_date_cols:
    orders[c] = pd.to_datetime(orders[c],errors = 'coerce')
items['shipping_limit_date'] = pd.to_datetime(items['shipping_limit_date'],errors = 'coerce')

print("Earliest Order Dates:",orders['order_purchase_timestamp'].min())
print("Latest Order dates:",orders['order_purchase_timestamp'].max())
```

-------

## Q6. Handle missing values

```python
products['product_category_name']=products['product_category_name'].fillna('unknown')
for c in ['product_name_lenght','product_description_lenght','product_photos_qty','product_weight_g','product_length_cm','product_height_cm','product_width_cm']:
    products[c] = products[c].fillna(products[c].median())
print("Missing Product Category:",products.product_category_name.isna().sum())
print("Missing Product Numeric Value:",products.select_dtypes(include = np.number).isna().sum().sum())
```

**Why:** Missing product category is retained as unknown because deleting those products would lose sales information. Small numeric gaps are replaced by the median because the median is less sensitive to extreme values. Delivery-date missing values are retained because they carry operational meaning (for example, the order may not have been delivered).

-------

## Q7. Check unusual values

```python
for c in ['price','freight_value']:
    print(f'{c}: negative = {(items[c]<0).sum()},zero = {(items[c]==0).sum()},99th Percentile={items[c].quantile(.99):.2f}, max = {items[c].max():.2f}')
print("Order-Items Rows (used as quantity because no explicit quantity column exists:",len(items))
print("Rows with Price > 99th Percentile:",(items.price > items.price.quantile(.99)).sum())
```

**Observation:** There are no negative prices or freight values. Freight has 383 zero values. Price is right-skewed; the maximum price is 6735. These high values are not automatically deleted because they may be legitimate products. The dataset has no separate quantity column; each order-item row represents one purchased item, so row count is used as item quantity.

# Prepare the master analysis table

```python
translation=pd.read_csv('product_category_name_translation.csv')
cat_map=translation.set_index('product_category_name')['product_category_name_english'].to_dict()
products['category_english']=products['product_category_name'].map(cat_map).fillna(products['product_category_name'].str.replace('_',' '))

master=(items.merge(orders[['order_id','customer_id','order_status','order_purchase_timestamp','order_delivered_customer_date','order_estimated_delivery_date']],on='order_id',how='left')
          .merge(customers[['customer_id','customer_unique_id','customer_state','customer_city']],on='customer_id',how='left')
          .merge(products[['product_id','product_category_name','category_english']],on='product_id',how='left'))
master['sales_value']=master['price']
master['delivery_days']=np.where(master.order_delivered_customer_date.notna(),(master.order_delivered_customer_date-master.order_purchase_timestamp).dt.total_seconds()/86400,np.nan)
master['delivered_late']=np.where(master.order_delivered_customer_date.notna(),master.order_delivered_customer_date>master.order_estimated_delivery_date,np.nan)
display(master.head())
```

## Q13. Top 10 products by quantity and sales

## Q14. Monthly sales trend

```python
monthly = (master.assign(month = master.order_purchase_timestamp.dt.to_period("M").astype(str)).groupby("month").sales_value.sum())
print("Highest Sales Month:",monthly.idxmax(),monthly.max())
print("Lowest Sales Month:",monthly.idxmin(),monthly.min())
monthly.plot(figsize = (10,5), marker = 'o',title = "Monthly sales Trend")
plt.xlabel("Month")
plt.ylabel("Product Sales")
plt.xticks(rotation = 60)
plt.show()
```

**Trend:** Sales rise strongly from 2017 into 2018, with a peak in November 2017. The first and last partial months should be interpreted carefully because the dataset does not contain complete monthly coverage at both ends.

# Part 6 — Intermediate EDA 
## Q18. Does product price affect quantity sold?

```python
pq = items.groupby("product_id").agg(average_price = ('price','mean'),quantity_sold = ("order_id","size")).reset_index()
corr = pq.average_price.corr(pq.quantity_sold)
print("Correlation:",corr)
```

```python
plt.figure(figsize = (8,5))
plt.scatter(pq.average_price,pq.quantity_sold,alpha= 0.35,s = 18)
plt.title("Product Price V/S Quantity Sold")
plt.xlabel("Average Product Price")
plt.ylabel("Quantity Sold")
plt.show()
```

**Interpretation:** The correlation is approximately **-0.032**, which is very close to zero. In this dataset there is little linear association between average product price and the number of units sold at product level. Correlation does not prove causation.

-----

## Q19. Delivery performance

```python
delivered = orders[orders.order_status == 'delivered'].dropna(subset = ['order_delivered_customer_date']).copy()
delivered['Delivery_Days'] = (delivered.order_delivered_customer_date - delivered.order_purchase_timestamp).dt.total_seconds()/86400
late = delivered.order_delivered_customer_date > delivered.order_estimated_delivery_date
print("Average Delivery days:",delivered.Delivery_Days.mean())
print("Minimum Delivery days:",delivered.Delivery_Days.min())
print("Maximum Delivery days:",delivered.Delivery_Days.max())
print("Median Delivery days:",delivered.Delivery_Days.median())
print("Delivered Order Analyse:",len(delivered))
print("Delivered later than estimated:",late.sum())
print("Late Percentage:",late.mean()*100)      
```

```python
plt.figure(figsize = (8,5))
plt.hist(delivered.Delivery_Days,bins = 40)
plt.title("Delivery Time Distribution")
plt.xlabel("Delivery Time (Days)")
plt.ylabel("Delivered Orders")
plt.show()
```
